```python
import numpy as np
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv("/content/Social_Network_Ads.csv")
```

```python
df=df.iloc[:,2:]
```

```python
df.sample(5)
```

#Train Test Split

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test =train_test_split(df.drop('Purchased',axis = 1),
                                                              df['Purchased'],test_size=0.3,random_state=0)
```

```python
X_train.shape ,X_test.shape
```

#Standard Scaler

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
#fiting a scaler for train data
scaler.fit(X_train)
#Transfer the train and test data
X_test_scaled =scaler.transform(X_train)
X_train_scaled = scaler.transform(X_test)
```

```python
scaler.mean_
```

```python
X_train_scaled=pd.DataFrame(X_train_scaled,columns=X_train.columns)
X_test_scaled=pd.DataFrame(X_test_scaled,columns=X_test.columns)
```

```python
np.round(X_train.describe(),1)
```

```python
np.round(X_train_scaled.describe(),1)
```

```python
fig , (ax1 , ax2)= plt.subplots(ncols=2 ,figsize=(12,5))
ax1.scatter(X_train['Age'],X_train['EstimatedSalary'])
ax1.set_title('Before Scaleing')
ax2.scatter(X_train_scaled['Age'],X_train_scaled['EstimatedSalary'],color='red')
ax2.set_title('After Scaleing')
plt.show()
```

```python
fig , (ax1 , ax2)= plt.subplots(ncols=2 ,figsize=(12,5))
ax1.set_title(' Age Before Scaleing')
sns.kdeplot(X_train['Age'],ax=ax1)


ax2.set_title(' Age After Scaleing')
sns.kdeplot(X_train_scaled['Age'],ax=ax2)
plt.show()
```

```python
fig , (ax1 , ax2)= plt.subplots(ncols=2 ,figsize=(12,5))
ax1.set_title(' EstimatedSalary Before Scaleing')
sns.kdeplot(X_train['EstimatedSalary'],ax=ax1)


ax2.set_title(' EstimatedSalary After Scaleing')
sns.kdeplot(X_train_scaled['EstimatedSalary'],ax=ax2)
plt.show()
```
