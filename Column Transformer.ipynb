```python
import numpy as np
import pandas as pd
```

```python
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder
```

```python
df= pd.read_csv('/content/covid_toy.csv')
```

```python
df.head()
```

```python
df['cough'].value_counts()
```

```python
from sklearn.model_selection import train_test_split
X_train , X_test , y_train , y_test = train_test_split(df.drop(columns=['has_covid']),df['has_covid'],test_size=0.2)
```

```python
from sklearn.compose import ColumnTransformer
```

```python
from pandas.core.arrays import sparse
transformer = ColumnTransformer(transformers=[("tnf1",SimpleImputer(),['fever']),
                                              ("tnf2",OrdinalEncoder(categories=[['Mild','Strong']]),['cough']),     #giving prority to 0 and 1 in cough column
                                              ("tnf3",OneHotEncoder(sparse_output=False,drop='first'),['gender','city'])
                                              ],remainder='passthrough')  ##we can drop or pass the values using remeinder
```

```python
transformer.fit_transform(X_train)
```

```python
transformer.fit_transform(X_train).shape
```

```python
transformer.transform(X_test)
```
